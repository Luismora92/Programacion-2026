# Problema #1: Calculadora de propina inteligente

## Problema

Calcular de forma automatica cuanto dinero se debe dejar de propina en un restaurante y cual es el monto final para pagar.

## Entrada

El total de la cuenta tener en cuenta que puede venir en numero decimal se usa float y el porcentaje de propina igual.

## Proceso

se multiplica el total de la cuenta por el porcentaje de la propina y se divide entre 100 para obtener el valor monetario de la propina y se suma el valor de la propina al total inicial.

## Salida esperada

El monto calculado de la propina y el total de la cuenta.

```python
total_cuenta = float(input("Ingrese el total de la cuenta: "))
porcentaje = float(input("Ingrese el porcentaje de propina: "))

propina = total_cuenta * (porcentaje / 100)
total_final = total_cuenta + propina

print(f"Monto de la propina: {propina}")
print(f"Total a pagar: {total_final}")
```

# Problema #2: Detector de número par/impar

## Problema

Saber si el numero que se ingresa sea par o impar para que se clasifique correctamente.

## Entrada

un numero entero ingresado por el usuario.

## Proceso

se usa el operador para dividir el numero entre 2 para saber si es par.

se usa el condicional if o else si el resultado es igual a 0 para saber si es par o impar

## Salida esperada

un mensaje en la pantalla que indique que el numero ingresado es par o impar.

```python
numero = int(input("Ingrese un número entero: "))

if numero % 2 == 0:
    print("El número es par")
else:
    print("El número es impar")
```

# Problema #3: Clasificador de edad

## Problema

categorizar a una persona segun su edad en diferentes etapas de la vida para llevar un control

## Entrada

la edad de la persona en años como un numero entero

## Proceso

se calcula la edad usando un condicional if-elif-else y se verifica si es menor de 19 o si es mayor a 65.

## Salida esperada

un mensaje que diga la clasificacion correspondiente de la persona .

```python
edad = int(input("Ingrese su edad: "))

if edad < 18:
    print("Es menor de edad")
elif edad < 65:
    print("Es adulto")
else:
    print("Es adulto mayor")
```

# Problema #4: Validador de compra con descuento

## Problema

validar si un cliente tiene derecho a recibir un descuento en la tienda si es cliente frecuente o si su compra supera un monto en especifico

## Entrada

el nombre del cliente, el monto total de la compra tambien tener en cuenta que pueden haber decimales y si el cliente es nuevo o no en la tienda

## Proceso

se calcula mediante la condicion or y revisa si el cliente es frecuente o no y si la compra supera los 75000 para que se le de el descuento de la tienda en el caso que cumpla una de las condiciones.

## Salida esperada

Un mensaje que diga si el cliente puede tener beneficio del descuento.

## Pruebas manuales


```python
cliente = input("Ingrese el nombre del cliente: ")
monto = float(input("Ingrese el monto de la compra: "))
frecuente = input("¿Es cliente frecuente? (escriba 'si' o 'no'): ")

# Condición: Se aplica descuento si es cliente frecuente O si el monto es mayor a 75000 (incluso si no es frecuente)
if frecuente == "si" or monto > 75000:
    print(f"¡Felicidades {cliente}! Su compra aplica para un descuento especial.")
else:
    print(f"Hola {cliente}, su compra no cumple con los requisitos para descuento.")
```
